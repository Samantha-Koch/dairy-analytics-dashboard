```python
from jupyter_dash import JupyterDash
from dash import dcc, html
import plotly.express as px
import pandas as pd

# Load the dashboard dataset from your modeling notebook
dashboard_df = pd.read_csv("dashboard_df.csv")

importance_nm = pd.read_csv("importance_nm.csv")
# Use your dashboard dataset
df = dashboard_df

# Build the app
app = JupyterDash(__name__)

# KPI values
avg_margin = df["net_margin"].mean()
avg_pred_margin = df["predicted_margin"].mean()
avg_milk_per_cow = df["milk_per_cow_annual"].mean()

# Layout
app.layout = html.Div([
    html.H1("Dairy Economics Dashboard", style={"textAlign": "center"}),

    dcc.Tabs([
        # -------------------------
        # TAB 1: OVERVIEW
        # -------------------------
        dcc.Tab(label="Overview", children=[
            
            # KPI Row
            html.Div([
                html.Div([
                    html.H3("Avg Net Margin"),
                    html.P(f"{avg_margin:.2f}")
                ], style={"width": "30%", "display": "inline-block"}),

                html.Div([
                    html.H3("Avg Predicted Margin"),
                    html.P(f"{avg_pred_margin:.2f}")
                ], style={"width": "30%", "display": "inline-block"}),

                html.Div([
                    html.H3("Milk per Cow (Avg)"),
                    html.P(f"{avg_milk_per_cow:.0f}")
                ], style={"width": "30%", "display": "inline-block"}),
            ], style={"textAlign": "center"}),

            # Feature Importance Chart
            dcc.Graph(
                figure=px.bar(
                    importance_nm,
                    x="importance",
                    y="feature",
                    orientation="h",
                    title="Feature Importance — Random Forest"
                )
            ),
        ]),

        # -------------------------
        # TAB 2: MODEL INSIGHTS
        # -------------------------
        dcc.Tab(label="Model Insights", children=[
            dcc.Graph(
                figure=px.scatter(
                    dashboard_df,
                    x="net_margin",
                    y="predicted_margin",
                    trendline="ols",
                    title="Predicted vs Actual Net Margin"
                )
            )
        ]),
    ])
])

# Run inline
app.run(mode="inline")
```
